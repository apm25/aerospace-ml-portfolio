```python
import pandas as pd
```

```python
df = pd.read_csv("adult.data.csv")
print(df)
```

```python
print((df[df['sex'] == "Male"].age).mean())
```

```python
#Bachelors percentage all methods
print(len(df[df['education'] == "Bachelors"].education)/len(df.education) * 100)
print(df['education'].value_counts(normalize="True")['Bachelors'] *100)
print((df['education'] == "Bachelors").mean() * 100)
```

```python
print(df[~df['education'].isin(['Bachelors','Masters','Doctorate'])]['salary'].value_counts(normalize="True")['>50K']*100)
```

```python
print(df['education'].value_counts()['Bachelors'])
```

```python
print(df[df['hours-per-week'] == 1] )
```

```python
min_work_hours = 1
print(df[df['hours-per-week'].isin([min_work_hours])]['salary'].value_counts(normalize="True")['>50K'])
```
